```python
# =====================
# 1. Import Libraries
# =====================
import os, cv2
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    cohen_kappa_score, roc_auc_score, classification_report,
    confusion_matrix
)
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.utils.multiclass import unique_labels
from skimage.feature import hog
import pandas as pd

# =====================
# 2. Load Dataset (with grayscale + visualization)
# =====================
def load_dataset(data_dir, img_size=64, max_per_class=300, show_samples=True):
    X, y, labels = [], [], []
    class_names = os.listdir(data_dir)
    for idx, cls in enumerate(class_names):
        folder = os.path.join(data_dir, cls)
        if not os.path.isdir(folder): 
            continue
        count = 0
        for file in os.listdir(folder):
            if count >= max_per_class: 
                break
            img_path = os.path.join(folder, file)
            img = cv2.imread(img_path)
            if img is None: 
                continue
            img = cv2.resize(img, (img_size, img_size))
            gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
            X.append(gray / 255.0)
            y.append(idx)
            count += 1
        labels.append(cls)

    X, y = np.array(X), np.array(y)

    # 🔹 Show first 10 samples (Original vs Grayscale)
    if show_samples:
        plt.figure(figsize=(15, 6))
        for i in range(10):
            # Original (reload for display only)
            img_path = os.path.join(data_dir, class_names[y[i]], os.listdir(os.path.join(data_dir, class_names[y[i]]))[0])
            orig = cv2.imread(img_path)
            orig = cv2.resize(orig, (img_size, img_size))
            gray = (X[i] * 255).astype("uint8")

            # Original
            plt.subplot(2, 10, i+1)
            plt.imshow(cv2.cvtColor(orig, cv2.COLOR_BGR2RGB))
            plt.title("Original", fontsize=8)
            plt.axis("off")

            # Grayscale
            plt.subplot(2, 10, i+11)
            plt.imshow(gray, cmap="gray")
            plt.title("Grayscale", fontsize=8)
            plt.axis("off")

        plt.suptitle("Original vs Grayscale (First 10 Images)", fontsize=14)
        plt.show()

    return X, y, labels


# =====================
# 3. Load Data
# =====================
data_dir = r"C:\Users\Rishitha\Downloads\archive\PlantVillage"
X, y, class_names = load_dataset(data_dir, show_samples=True)
print("Dataset:", X.shape, "Labels:", len(class_names))

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

# =====================
# 4. HOG Feature Extraction
# =====================
def extract_hog(images, show_example=True):
    feats, hog_images = [], []
    for i, img in enumerate(images):
        feat, hog_img = hog(img, orientations=9, pixels_per_cell=(8,8),
                            cells_per_block=(2,2), visualize=True)
        feats.append(feat)
        if i < 5:  # store first 5 HOG visualizations
            hog_images.append((img, hog_img))
    feats = np.array(feats)

    # 🔹 Show HOG visualization
    if show_example:
        plt.figure(figsize=(12, 6))
        for i, (g, h) in enumerate(hog_images):
            plt.subplot(2, 5, i+1)
            plt.imshow(g, cmap="gray")
            plt.title("Grayscale", fontsize=8)
            plt.axis("off")

            plt.subplot(2, 5, i+6)
            plt.imshow(h, cmap="gray")
            plt.title("HOG", fontsize=8)
            plt.axis("off")
        plt.suptitle("HOG Feature Visualization (5 samples)", fontsize=14)
        plt.show()

    return feats

X_train_hog, X_test_hog = extract_hog(X_train), extract_hog(X_test)

# =====================
# 5. Classifiers
# =====================
classifiers = {
    "LogReg": LogisticRegression(max_iter=200),
    "KNN": KNeighborsClassifier(n_neighbors=5),
    "DecisionTree": DecisionTreeClassifier(),
    "RandomForest": RandomForestClassifier(n_estimators=100),
    "SVM": SVC(probability=True)
}

results = []

# =====================
# 6. Evaluation
# =====================
def evaluate(X_tr, X_te, y_tr, y_te, feature_name):
    for clf_name, clf in classifiers.items():
        clf.fit(X_tr, y_tr)
        y_pred = clf.predict(X_te)

        acc = accuracy_score(y_te, y_pred)
        prec = precision_score(y_te, y_pred, average="macro")
        rec = recall_score(y_te, y_pred, average="macro")
        f1 = f1_score(y_te, y_pred, average="macro")
        kappa = cohen_kappa_score(y_te, y_pred)

        try:
            y_prob = clf.predict_proba(X_te)
            roc = roc_auc_score(y_te, y_prob, multi_class="ovr")
        except:
            roc = np.nan

        results.append([feature_name, clf_name, acc, prec, rec, f1, kappa, roc])

        print(f"\n🔹 {feature_name} + {clf_name}")
        print(f"Acc={acc:.4f}, Prec={prec:.4f}, Rec={rec:.4f}, "
              f"F1={f1:.4f}, Kappa={kappa:.4f}, ROC={roc:.4f}")

        # ✅ Auto-handle class mismatch
        labels_used = unique_labels(y_te, y_pred)
        print(classification_report(
            y_te, y_pred,
            labels=labels_used,
            target_names=[class_names[i] for i in labels_used],
            zero_division=0
        ))

        # Confusion Matrix
        cm = confusion_matrix(y_te, y_pred, labels=labels_used)
        plt.figure(figsize=(8,6))
        sns.heatmap(cm, cmap="Blues", annot=True,
                    xticklabels=[class_names[i] for i in labels_used],
                    yticklabels=[class_names[i] for i in labels_used])
        plt.title(f"{feature_name} + {clf_name}")
        plt.show()


# =====================
# 7. Run Evaluation (HOG)
# =====================
evaluate(X_train_hog, X_test_hog, y_train, y_test, "HOG")

# =====================
# 8. Results Summary
# =====================
df_results = pd.DataFrame(results, columns=["Feature","Classifier","Accuracy","Precision","Recall","F1","Kappa","ROC-AUC"])
print("\n=== Final Results ===")
print(df_results)
```

```python
# =====================
# Imports
# =====================
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    cohen_kappa_score, roc_auc_score, classification_report,
    confusion_matrix
)
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.utils.multiclass import unique_labels
from tensorflow.keras.applications import VGG16
from tensorflow.keras.applications.vgg16 import preprocess_input
from tensorflow.keras.preprocessing.image import img_to_array

# =====================
# Load Dataset
# =====================
def load_dataset(data_dir, img_size=64, max_per_class=300):
    X, y, labels = [], [], []
    class_names = os.listdir(data_dir)
    for idx, cls in enumerate(class_names):
        folder = os.path.join(data_dir, cls)
        if not os.path.isdir(folder): 
            continue
        count = 0
        for file in os.listdir(folder):
            if count >= max_per_class: break
            img_path = os.path.join(folder, file)
            img = cv2.imread(img_path)
            if img is None: continue
            img = cv2.resize(img, (img_size, img_size))
            gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
            X.append(gray / 255.0)  # normalize
            y.append(idx)
            count += 1
        labels.append(cls)
    return np.array(X), np.array(y), labels

data_dir = r"C:\Users\Rishitha\Downloads\archive\PlantVillage"
X, y, class_names = load_dataset(data_dir)
print("Dataset:", X.shape, "Labels:", len(class_names))

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

# =====================
# Show some grayscale images
# =====================
plt.figure(figsize=(12,6))
for i in range(10):
    plt.subplot(2,5,i+1)
    plt.imshow(X_train[i], cmap="gray")
    plt.title(class_names[y_train[i]])
    plt.axis("off")
plt.suptitle("Sample Grayscale Images", fontsize=16)
plt.show()

# =====================
# Extract VGG16 features
# =====================
def extract_vgg_features(images):
    vgg = VGG16(weights="imagenet", include_top=False, input_shape=(64,64,3))
    feats = []
    for img in images:
        # Grayscale to RGB
        img_rgb = cv2.cvtColor((img * 255).astype(np.uint8), cv2.COLOR_GRAY2RGB)
        arr = cv2.resize(img_rgb, (64, 64))
        arr = img_to_array(arr)
        arr = np.expand_dims(arr, axis=0)
        arr = preprocess_input(arr)
        feat = vgg.predict(arr, verbose=0)
        feats.append(feat.flatten())
    return np.array(feats)

X_train_vgg = extract_vgg_features(X_train)
X_test_vgg = extract_vgg_features(X_test)

# =====================
# Classifiers
# =====================
classifiers = {
    "LogReg": LogisticRegression(max_iter=200),
    "KNN": KNeighborsClassifier(n_neighbors=5),
    "DecisionTree": DecisionTreeClassifier(),
    "RandomForest": RandomForestClassifier(n_estimators=100),
    "SVM": SVC(probability=True)
}

# =====================
# Evaluation function
# =====================
def evaluate(X_tr, X_te, y_tr, y_te, feature_name):
    for clf_name, clf in classifiers.items():
        clf.fit(X_tr, y_tr)
        y_pred = clf.predict(X_te)

        acc = accuracy_score(y_te, y_pred)
        prec = precision_score(y_te, y_pred, average="macro")
        rec = recall_score(y_te, y_pred, average="macro")
        f1 = f1_score(y_te, y_pred, average="macro")
        kappa = cohen_kappa_score(y_te, y_pred)

        try:
            y_prob = clf.predict_proba(X_te)
            roc = roc_auc_score(y_te, y_prob, multi_class="ovr")
        except:
            roc = np.nan

        print(f"\n🔹 {feature_name} + {clf_name}")
        print(f"Acc={acc:.4f}, Prec={prec:.4f}, Rec={rec:.4f}, F1={f1:.4f}, Kappa={kappa:.4f}, ROC={roc:.4f}")

        labels_used = unique_labels(y_te, y_pred)
        print(classification_report(
            y_te, y_pred,
            labels=labels_used,
            target_names=[class_names[i] for i in labels_used],
            zero_division=0
        ))

        cm = confusion_matrix(y_te, y_pred, labels=labels_used)
        plt.figure(figsize=(8,6))
        sns.heatmap(cm, cmap="Blues", annot=True,
                    xticklabels=[class_names[i] for i in labels_used],
                    yticklabels=[class_names[i] for i in labels_used])
        plt.title(f"{feature_name} + {clf_name}")
        plt.show()

# =====================
# Run Evaluation on VGG16 features
# =====================
evaluate(X_train_vgg, X_test_vgg, y_train, y_test, "VGG16")
```

```python
# =====================
# 9. Improved CNN Training (End-to-End)
# =====================
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, GlobalAveragePooling2D, Dense, Dropout, BatchNormalization
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report

# Set random seeds for reproducibility
np.random.seed(42)
tf.random.set_seed(42)

# Preprocess: CNN expects (H, W, C)
X_cnn = np.expand_dims(X, axis=-1)              # add channel for grayscale
X_cnn = np.repeat(X_cnn, 3, axis=-1)            # convert to 3 channels
y_cnn = to_categorical(y, num_classes=len(class_names))

# Train/validation split
from sklearn.model_selection import train_test_split
X_train_cnn, X_val_cnn, y_train_cnn, y_val_cnn = train_test_split(
    X_cnn, y_cnn, test_size=0.2, stratify=y, random_state=42
)
print("CNN data shapes:", X_train_cnn.shape, y_train_cnn.shape)

# =====================
# Data Augmentation
# =====================
datagen = ImageDataGenerator(
    rotation_range=20,
    width_shift_range=0.1,
    height_shift_range=0.1,
    zoom_range=0.1,
    horizontal_flip=True
)
datagen.fit(X_train_cnn)

# =====================
# CNN Architecture (Improved)
# =====================
cnn = Sequential([
    Conv2D(32, (3,3), activation='relu', input_shape=(64,64,3)),
    BatchNormalization(),
    MaxPooling2D(2,2),

    Conv2D(64, (3,3), activation='relu'),
    BatchNormalization(),
    MaxPooling2D(2,2),

    Conv2D(128, (3,3), activation='relu'),
    BatchNormalization(),
    MaxPooling2D(2,2),

    GlobalAveragePooling2D(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(len(class_names), activation='softmax')
])

cnn.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy', tf.keras.metrics.Precision(), tf.keras.metrics.Recall(), tf.keras.metrics.AUC()]
)

# =====================
# Training with EarlyStopping
# =====================
es = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)

history = cnn.fit(
    datagen.flow(X_train_cnn, y_train_cnn, batch_size=32),
    validation_data=(X_val_cnn, y_val_cnn),
    epochs=30,
    callbacks=[es],
    verbose=1
)

# =====================
# Evaluation
# =====================
cnn_eval = cnn.evaluate(X_val_cnn, y_val_cnn, verbose=0)
print(f"\n🔹 CNN Validation Accuracy = {cnn_eval[1]:.4f}")
print(f"🔹 CNN Precision = {cnn_eval[2]:.4f}, Recall = {cnn_eval[3]:.4f}, AUC = {cnn_eval[4]:.4f}")

# =====================
# Plot Training Curves
# =====================
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.plot(history.history['accuracy'], label="Train Acc")
plt.plot(history.history['val_accuracy'], label="Val Acc")
plt.legend(); plt.title("CNN Accuracy")

plt.subplot(1,2,2)
plt.plot(history.history['loss'], label="Train Loss")
plt.plot(history.history['val_loss'], label="Val Loss")
plt.legend(); plt.title("CNN Loss")
plt.show()

# =====================
# Confusion Matrix + Report
# =====================
y_pred_cnn = np.argmax(cnn.predict(X_val_cnn), axis=1)
y_true_cnn = np.argmax(y_val_cnn, axis=1)

unique_labels = np.unique(y_true_cnn)
class_names_fixed = [class_names[i] for i in unique_labels]

cm = confusion_matrix(y_true_cnn, y_pred_cnn, labels=unique_labels)
plt.figure(figsize=(8,6))
sns.heatmap(cm, cmap="Blues", xticklabels=class_names_fixed, yticklabels=class_names_fixed, annot=True, fmt="d")
plt.title("CNN Confusion Matrix")
plt.xlabel("Predicted"); plt.ylabel("True")
plt.show()

print("\nCNN Classification Report:")
print(classification_report(
    y_true_cnn, y_pred_cnn,
    labels=unique_labels,
    target_names=class_names_fixed,
    zero_division=0
))
```

```python
# === Compare ML (HOG, LBP) vs CNN (end-to-end) on PlantVillage ===
# Requirements: opencv-python, scikit-image, scikit-learn, tensorflow, matplotlib, seaborn, pandas
# pip install opencv-python scikit-image scikit-learn tensorflow matplotlib seaborn pandas

import os, cv2, random, time
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from collections import OrderedDict

from skimage.feature import hog, local_binary_pattern
from skimage import filters

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    classification_report, confusion_matrix
)
from sklearn.utils.multiclass import unique_labels

import tensorflow as tf
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, BatchNormalization

# ---------------------------
# User parameters
# ---------------------------
DATA_DIR = r"C:\Users\Rishitha\Downloads\archive\PlantVillage"  # <- your folder
IMG_SIZE = 128      # image size for CNN and feature extraction
MAX_PER_CLASS = 500 # cap for speed; set None for all
TEST_SIZE = 0.2
RANDOM_STATE = 42

# ---------------------------
# 1) Load dataset into arrays (RGB and Grayscale)
# ---------------------------
def load_dataset_as_arrays(folder, img_size=128, max_per_class=None):
    X_rgb, X_gray, y, class_names = [], [], [], []
    dirs = [d for d in os.listdir(folder) if os.path.isdir(os.path.join(folder, d))]
    dirs.sort()
    for idx, cls in enumerate(dirs):
        class_folder = os.path.join(folder, cls)
        files = [f for f in os.listdir(class_folder) if os.path.isfile(os.path.join(class_folder, f))]
        if max_per_class:
            files = files[:max_per_class]
        count = 0
        for f in files:
            path = os.path.join(class_folder, f)
            img_bgr = cv2.imread(path)
            if img_bgr is None:
                continue
            img_bgr = cv2.resize(img_bgr, (img_size, img_size))
            img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
            img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
            X_rgb.append(img_rgb)
            X_gray.append(img_gray.astype(np.float32) / 255.0)
            y.append(idx)
            count += 1
        # add class name regardless (we'll filter later)
        class_names.append(cls)
    X_rgb = np.array(X_rgb)
    X_gray = np.array(X_gray)
    y = np.array(y)
    return X_rgb, X_gray, y, class_names

print("Loading dataset (this may take a while)...")
X_rgb, X_gray, y, class_names = load_dataset_as_arrays(DATA_DIR, IMG_SIZE, MAX_PER_CLASS)
print("Loaded:", X_rgb.shape, X_gray.shape, "labels:", y.shape, "declared classes:", len(class_names))

# ---------------------------
# Quick check & fix classes (drop any class names that have zero samples)
# ---------------------------
# Count samples per declared class index
counts = OrderedDict()
for i,c in enumerate(class_names):
    counts[c] = int((y==i).sum())
print("Samples per declared class (showing classes with >0):")
for c,n in counts.items():
    if n>0:
        print(f" - {c}: {n}")
# Build mapping only for classes actually present
present_label_indices = sorted(list(set(y)))
class_names_fixed = [class_names[i] for i in present_label_indices]
print("\nUsing %d classes actually present." % len(class_names_fixed))

# Re-map labels to a contiguous range 0..C-1 (in case some class indices missing)
old_to_new = {old:i for i,old in enumerate(present_label_indices)}
y_mapped = np.array([old_to_new[val] for val in y])
class_names_final = [class_names[i] for i in present_label_indices]

# ---------------------------
# Train/test split (same for ML)
# ---------------------------
Xg_train, Xg_test, Xr_train, Xr_test, y_train_raw, y_test_raw = train_test_split(
    X_gray, X_rgb, y_mapped, test_size=TEST_SIZE, stratify=y_mapped, random_state=RANDOM_STATE
)
print("Train/test sizes:", Xg_train.shape[0], Xg_test.shape[0])

# ---------------------------
# Feature extractors for ML
# ---------------------------
def extract_hog_features(X_gray_arr, pixels_per_cell=(8,8), cells_per_block=(2,2)):
    feats = [hog(img,
                 orientations=9,
                 pixels_per_cell=pixels_per_cell,
                 cells_per_block=cells_per_block,
                 visualize=False,
                 feature_vector=True) for img in X_gray_arr]
    return np.array(feats)

def extract_lbp_features(X_gray_arr, P=8, R=1):
    feats = []
    for img in X_gray_arr:
        lbp = local_binary_pattern(img, P=P, R=R, method="uniform")
        hist, _ = np.histogram(lbp.ravel(), bins=np.arange(0, P+3), density=True) # P+2 + 1 edge
        feats.append(hist)
    return np.array(feats)

# HOG + LBP features
print("Extracting HOG features...")
Xtr_hog = extract_hog_features(Xg_train)
Xte_hog = extract_hog_features(Xg_test)
print("HOG shapes:", Xtr_hog.shape, Xte_hog.shape)

print("Extracting LBP features...")
Xtr_lbp = extract_lbp_features(Xg_train)
Xte_lbp = extract_lbp_features(Xg_test)
print("LBP shapes:", Xtr_lbp.shape, Xte_lbp.shape)

# ---------------------------
# ML Models training & evaluation
# ---------------------------
ml_models = {
    "LogisticRegression": make_pipeline(StandardScaler(with_mean=False), LogisticRegression(max_iter=500)),
    "KNN": KNeighborsClassifier(n_neighbors=5),
    "DecisionTree": DecisionTreeClassifier(random_state=RANDOM_STATE),
    "RandomForest": RandomForestClassifier(n_estimators=150, random_state=RANDOM_STATE)
}

def evaluate_and_report(y_true, y_pred, model_name):
    acc = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred, average='macro', zero_division=0)
    rec = recall_score(y_true, y_pred, average='macro', zero_division=0)
    f1 = f1_score(y_true, y_pred, average='macro', zero_division=0)
    print(f"\n-- {model_name} --")
    print(f"Accuracy={acc:.4f} Precision={prec:.4f} Recall={rec:.4f} F1={f1:.4f}")
    # safe report: only labels present
    labels_used = unique_labels(y_true, y_pred)
    names_used = [class_names_final[i] for i in labels_used]
    print(classification_report(y_true, y_pred, labels=labels_used, target_names=names_used, zero_division=0))
    cm = confusion_matrix(y_true, y_pred, labels=labels_used)
    plt.figure(figsize=(8,6))
    sns.heatmap(cm, cmap="Blues", xticklabels=names_used, yticklabels=names_used, annot=False)
    plt.title(f"Confusion Matrix: {model_name}")
    plt.xlabel("Predicted"); plt.ylabel("True"); plt.show()
    return dict(Accuracy=acc, Precision=prec, Recall=rec, F1=f1)

results = []

# Evaluate models on HOG features
for name, model in ml_models.items():
    print("\nTraining on HOG:", name)
    model.fit(Xtr_hog, y_train_raw)
    ypred = model.predict(Xte_hog)
    r = evaluate_and_report(y_test_raw, ypred, f"HOG + {name}")
    r['Method'] = f"HOG + {name}"
    results.append(r)

# Evaluate models on LBP features
for name, model in ml_models.items():
    print("\nTraining on LBP:", name)
    model.fit(Xtr_lbp, y_train_raw)
    ypred = model.predict(Xte_lbp)
    r = evaluate_and_report(y_test_raw, ypred, f"LBP + {name}")
    r['Method'] = f"LBP + {name}"
    results.append(r)

# ---------------------------
# Simple CNN (end-to-end) training on RGB images
# ---------------------------
# Prepare data
Xc = np.array(Xr_train, dtype=np.float32) / 255.0   # train RGB (0..1)
Xc_test = np.array(Xr_test, dtype=np.float32) / 255.0
yc = to_categorical(y_train_raw, num_classes=len(class_names_final))
yc_test = to_categorical(y_test_raw, num_classes=len(class_names_final))

# Build simple CNN
def build_simple_cnn(input_shape=(IMG_SIZE,IMG_SIZE,3), num_classes=len(class_names_final)):
    model = Sequential([
        Conv2D(32, (3,3), activation='relu', input_shape=input_shape),
        BatchNormalization(),
        MaxPooling2D(2,2),

        Conv2D(64, (3,3), activation='relu'),
        BatchNormalization(),
        MaxPooling2D(2,2),

        Conv2D(128, (3,3), activation='relu'),
        BatchNormalization(),
        MaxPooling2D(2,2),

        Flatten(),
        Dense(256, activation='relu'),
        Dropout(0.5),
        Dense(num_classes, activation='softmax')
    ])
    return model

cnn = build_simple_cnn((IMG_SIZE,IMG_SIZE,3), len(class_names_final))
cnn.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
cnn.summary()

# Train CNN (use validation split)
es = tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
history = cnn.fit(
    Xc, yc,
    validation_split=0.15,
    epochs=30,
    batch_size=32,
    callbacks=[es],
    verbose=1
)

# Evaluate CNN on the same ML test set (fair comparison)
y_pred_probs = cnn.predict(Xc_test)
y_pred_cnn = np.argmax(y_pred_probs, axis=1)
y_true_cnn = y_test_raw

rcnn = evaluate_and_report(y_true_cnn, y_pred_cnn, "CNN (end-to-end)")
rcnn['Method'] = 'CNN (end-to-end)'
results.append(rcnn)

# ---------------------------
# Summary plot and table
# ---------------------------
df = pd.DataFrame(results)[['Method','Accuracy','Precision','Recall','F1']]
print("\n=== Summary Table ===")
print(df.sort_values('Accuracy', ascending=False).to_string(index=False))

plt.figure(figsize=(10,5))
plt.title("Accuracy: ML models (HOG/LBP) vs CNN")
plt.bar(df['Method'], df['Accuracy'])
plt.xticks(rotation=30, ha='right')
plt.ylabel("Accuracy")
plt.ylim(0,1.0)
plt.tight_layout()
plt.show()
```

```python
import os
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.model_selection import train_test_split
import seaborn as sns

# =====================
# Load Dataset from Folder
# =====================
data_dir = r"C:\Users\Rishitha\Downloads\archive\PlantVillage"  # Update path if needed

img_size = (64, 64)
batch_size = 32

ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    image_size=img_size,
    batch_size=batch_size,
    label_mode='int',
    validation_split=0.2,
    subset='both',
    seed=42
)

class_names = ds[0].class_names

# Normalize the pixel values (0-1)
normalization_layer = tf.keras.layers.Rescaling(1./255)
train_ds = ds[0].map(lambda x, y: (normalization_layer(x), y))
val_ds = ds[1].map(lambda x, y: (normalization_layer(x), y))

train_ds = train_ds.cache().shuffle(1000).prefetch(buffer_size=tf.data.AUTOTUNE)
val_ds = val_ds.cache().prefetch(buffer_size=tf.data.AUTOTUNE)

# =====================
# CNN Model
# =====================
cnn_model = tf.keras.Sequential([
    tf.keras.layers.Conv2D(32, (3,3), activation='relu', input_shape=(64, 64, 3)),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(64, (3,3), activation='relu'),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(128, (3,3), activation='relu'),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(len(class_names), activation='softmax')
])

cnn_model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])

# =====================
# Train Model
# =====================
history = cnn_model.fit(train_ds, validation_data=val_ds, epochs=5)

# =====================
# Evaluate Model
# =====================
val_images = []
val_labels = []
for imgs, labels in val_ds:
    val_images.append(imgs)
    val_labels.append(labels)

val_images = tf.concat(val_images, axis=0)
val_labels = tf.concat(val_labels, axis=0)

pred_probs = cnn_model.predict(val_images)
y_pred = np.argmax(pred_probs, axis=1)

# =====================
# Classification Report & Confusion Matrix
# =====================
print("\nClassification Report:")
print(classification_report(val_labels, y_pred, target_names=class_names, zero_division=0))

print("\nConfusion Matrix:")
cm = confusion_matrix(val_labels, y_pred)
plt.figure(figsize=(12, 8))
sns.heatmap(cm, annot=True, fmt="d", xticklabels=class_names, yticklabels=class_names, cmap="Blues")
plt.title("CNN Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("True")
plt.show()
```
